```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.service import Service
import time
import os
```

```python
# 🔹 Path to chromedriver
service = Service("C:/chromedriver/chromedriver.exe")

# 🔹 Set download directory
download_path = os.path.abspath(r"C:\Users\LENOVO\OneDrive\Desktop\Cricsheet - Analysis\data_raw")

options = webdriver.ChromeOptions()
prefs = {
    "download.default_directory": download_path,
    "download.prompt_for_download": False,
    "directory_upgrade": True
}
options.add_experimental_option("prefs", prefs)

driver = webdriver.Chrome(options=options)
```

```python
# 🔹 Start browser
driver = webdriver.Chrome(service=service, options=options)

print("Opening Cricsheet...")
driver.get("https://cricsheet.org/matches/")
time.sleep(5)

# 🔹 Find all links
links = driver.find_elements(By.TAG_NAME, "a")

# 🔹 Download JSON zip files
for link in links:
    href = link.get_attribute("href")
    if href and "json.zip" in href:
        print("Downloading:", href)
        driver.get(href)
        time.sleep(5)

print("Download Completed")
driver.quit()
```
