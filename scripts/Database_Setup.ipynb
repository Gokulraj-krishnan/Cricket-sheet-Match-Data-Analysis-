```python
import sqlite3
import pandas as pd

# Create database
conn = sqlite3.connect("cricket_analytics.db")

print("Database created successfully")
```

```python
matches_df = pd.read_csv(r"C:\Users\LENOVO\OneDrive\Desktop\Cricsheet - Analysis\data_processed\matches.csv")
deliveries_df = pd.read_csv(r"C:\Users\LENOVO\OneDrive\Desktop\Cricsheet - Analysis\data_processed\deliveries.csv")

print(matches_df.shape)
print(deliveries_df.shape)
```

```python
matches_df.to_sql("matches", conn, if_exists="replace", index=False)
deliveries_df.to_sql("deliveries", conn, if_exists="replace", index=False)

print("Data loaded into SQL tables")
```

```python
cursor = conn.cursor()

cursor.execute("SELECT name FROM sqlite_master WHERE type='table';")
print(cursor.fetchall())
```

```python
query = """
SELECT season, COUNT(*) as total_matches
FROM matches
GROUP BY season
ORDER BY season;
"""

result = pd.read_sql(query, conn)
print(result)
```
